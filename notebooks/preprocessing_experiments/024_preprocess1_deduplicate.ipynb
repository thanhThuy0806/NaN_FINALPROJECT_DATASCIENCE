# Remove duplicated values

```python
import pandas as pd
```

```python
def deduplicate_steam_data(input_path: str, output_path: str):
    df = pd.read_csv(input_path)
    initial_count = len(df)
    print(f"Initial record count: {initial_count}")

    # Drop exact identical rows
    df.drop_duplicates(inplace=True)
    exact_drop_count = initial_count - len(df)
    print(f"Exact duplicates removed: {exact_drop_count}")

    # Drop Appid duplicates (Primary Key)
    # Sorting by PositiveReview (descending) ensures we keep the most established/populated record
    df.sort_values(by=['Appid', 'PositiveReview'], ascending=[True, False], inplace=True)
    df.drop_duplicates(subset=['Appid'], keep='first', inplace=True)
    appid_drop_count = (initial_count - exact_drop_count) - len(df)
    print(f"Appid duplicates removed: {appid_drop_count}")

    # Drop Name duplicates
    # Different Appids can sometimes share the exact same Name (e.g., Regional variants, GOTY editions, or bad Kaggle scrapes)
    # Keeping the most reviewed version prevents data leakage between train/test sets
    df.sort_values(by=['Name', 'PositiveReview'], ascending=[True, False], inplace=True)
    df.drop_duplicates(subset=['Name'], keep='first', inplace=True)
    name_drop_count = (initial_count - exact_drop_count - appid_drop_count) - len(df)
    print(f"Name duplicates removed: {name_drop_count}")

    print(f"Final record count: {len(df)}")
    print(f"Total duplicates removed: {initial_count - len(df)}")

    # Save the finalized dataset
    df.to_csv(output_path, index=False)


deduplicate_steam_data('../../data/output/SteamGames_Clean_2.csv', '../../data/output/SteamGames_Clean_3.csv')
```
